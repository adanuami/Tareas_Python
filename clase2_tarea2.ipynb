```python
estados=["Jalisco","Puebla","CDMX","Nuevo León","Oaxaca","Chiapas","Veracruz","Yucatán","Sonora","Jalisco","Puebla","Baja California","Querétaro","CDMX"]
```

```python
print("Lista inicial:",estados)
```

```python
print("Lista ordenada alfabeticamente:\n")
estados.sort()
print(estados)
```

```python
print("Despues de eliminar estados repetidos:\n")
estados.remove('CDMX')
estados.remove('Jalisco')
estados.remove('Puebla')
print(estados)
```

```python
print("Agregar 'Aguascalientes'al inicio:\n")
estados.insert(0,'Aguascalientes')
print(estados)
```

```python
print("Agregar 'Zacatecas'al final:\n")
estados.append('Zacatecas')
print(estados)
```

```python
print("¿'Oaxaca' esta en la lista?")
v='Oaxaca' in estados
if v==True:
    print("\n 'Oaxaca' si esta en la lista")
else:
    print("\n 'Oaxaca' no esta en la lista")
```

```python
print("La lista final es: \n")
print(estados)
longitud=len(estados)
print(f"\ncon una longitud de: {longitud} estados")
```
