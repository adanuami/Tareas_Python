```python
import numpy as np
```

```python
print("Definimos la matriz A y el vector V") 
```

```python
A=np.array([[2,4,6],[4,5,6],[3,1,-2]])
V=np.array([18,24,4])
```

```python
print("Matriz de coeficientes A:")
print(A)
print("\nVector de resultados V:", V)
print("\n")
```

```python
print("Determinante principal")
```

```python
D=np.linalg.det(A)
```

```python
print("Matrices modificadas")
```

```python
A1 = A.copy()
A2 = A.copy()
A3 = A.copy()
A1[:, 0] = V
A2[:, 1] = V
A3[:, 2] = V
```

```python
print("verificacion visual de matrices")
print("\nMatriz A1:")
print(A1)
print("\nMatriz A2:")
print(A2)
print("\nMatriz A3:")
print(A3)
```

```python
print("calculamos determinantes de matrices A1, A2 y A3")
```

```python
D1=np.linalg.det(A1)
D2=np.linalg.det(A2)
D3=np.linalg.det(A3)
```

```python
print("Finalmente calculamos x1=D1/D, x2=D2/D y x3=D3/D")
```

```python
x1=D1/D
x2=D2/D
x3=D3/D
```

```python
print("Los resultados son:")
print(f"\nx1= {round(x1,2)}")
print(f"\nx2= {round(x2,2)}")
print(f"\nx3= {round(x3,2)}")
```
