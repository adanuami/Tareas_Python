```python
import numpy as np
```

```python
u=np.array([2,4,6])
v=np.array([1,3,5])
```

```python
def distancia(u,v):
    return np.linalg.norm(v-u)

def normalizar(x):
    return x / np.linalg.norm(x)
u_normalizado=normalizar(u)
v_normalizado=normalizar(v)


escalar=np.dot(u,v)

norma_u=np.linalg.norm(u)
norma_v=np.linalg.norm(v)
```

```python
print(f"\nLa distancia entre u y v es: {round(distancia(u, v), 2)}")
print(f"\nEl vector u normalizado es: {np.round(u_normalizado, 3)}")
print(f"\nEl vector v normalizado es: {np.round(v_normalizado, 3)}")
print(f"\nEl resultado del producto escalar de u.v es: {round(escalar, 3)}")
```
